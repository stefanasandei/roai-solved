```python
import pandas as pd
import numpy as np
import re

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import f1_score
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.linear_model import Ridge
from sklearn.decomposition import TruncatedSVD
```

```python
root_path = "/home/stefan/ioai-prep/kits/riki"
seed = 42
```

# Data

```python
def clean_text(text_en: str):
    text_en = text_en.lower()
    text_en = re.sub(r"[^\w\s]", "", text_en)
    return text_en
```

```python
df_train = pd.read_csv(f"{root_path}/train_data.csv")
df_test = pd.read_csv(f"{root_path}/test_data.csv")

df_test_queries = pd.read_csv(f"{root_path}/test_queries.csv")
df_test_gallery = pd.read_csv(f"{root_path}/test_gallery.csv")

df_train["text_en"] = df_train["text_en"].apply(clean_text)
df_test["text_en"] = df_test["text_en"].apply(clean_text)

df_train.head()
```

# Subtask 1

```python
tfidf = TfidfVectorizer(ngram_range=(1, 2) , min_df=2, max_features=30000)

X_train_tfidf = tfidf.fit_transform(df_train["text_en"][df_train["is_food"] != -1])
test_feats = tfidf.transform(df_test["text_en"])

food_label = df_train["is_food"][df_train["is_food"] != -1]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_train_tfidf, food_label, stratify=food_label, test_size=0.2, random_state=seed)
```

```python
def evaluate(model):
    cv = cross_val_score(model, X_train, y_train, scoring='f1_macro', cv=3, n_jobs=-1)
    cv = cv.mean() - cv.std()

    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    f1 = f1_score(y_test, preds, average='macro')
    return cv, f1
```

```python
svc = LinearSVC(class_weight='balanced')

evaluate(svc)
```

```python
model = svc

model.fit(X_train, y_train)
subtask1 = model.predict(test_feats)
```

```python
pd.Series(subtask1).value_counts()
```

# Subtask 2

## Train mapper

```python
tfidf = TfidfVectorizer(ngram_range=(1, 4), min_df=2, analyzer='char', max_features=30000)

X_train_en = tfidf.fit_transform(df_train["text_en"][df_train["is_food"] == -1])
X_train_es = tfidf.transform(df_train["text_es"][df_train["is_food"] == -1])
```

```python
svd_en = TruncatedSVD(n_components=300)
svd_es = TruncatedSVD(n_components=300)

Z_train_en = svd_en.fit_transform(X_train_en)
Z_train_es = svd_es.fit_transform(X_train_es)
```

```python
mapper = Ridge(alpha=1.0)

mapper.fit(Z_train_es, Z_train_en)
```

## Retrieval

```python
X_query = tfidf.transform(df_test_queries["query_text_es"])
Z_query = svd_es.transform(X_query)
```

```python
Z_query_proj = mapper.predict(Z_query)
```

```python
X_gallery = tfidf.transform(df_test_gallery["text_en"])
Z_gallery = svd_en.transform(X_gallery)

Z_query_proj.shape, Z_gallery.shape
```

```python
sims = cosine_similarity(Z_query_proj, Z_gallery) # (2000, 3320)
indices = sims.argmax(axis=1)
indices.shape
```

```python
subtask2 = df_test_gallery.iloc[indices]["candidate_id"].values
```

# Submission

```python
def build_subtask(sid, ans):
    return pd.DataFrame({
        "subtaskID": sid,
        "datapointID": df_test_queries["query_id"] if sid == 2 else df_test["datapoint_id"],
        "answer": ans,
    })

subtasks = [
    (1, subtask1),
    (2, subtask2)
]

submission = pd.concat([build_subtask(sid, ans) for sid, ans in subtasks])
```
