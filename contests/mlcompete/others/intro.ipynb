```python
import pandas as pd

def celsius_to_fahrenheit(c_list):
    return [round(c * 9 / 5 + 32, 2) for c in c_list]

root_path = "/home/stefan/ioai-prep/kits/dataset-8f00ba3c-3f0a-4118-8c9c-412c6477275e"
test = pd.read_csv(f"{root_path}/test.csv")

test["temperature_f"] = celsius_to_fahrenheit(test["temperature_c"].tolist())

test[["SampleID", "temperature_f"]].to_csv(
    "submission.csv", index=False, float_format="%.2f"
)
```
